```python
import pandas as pd
df= pd.read_csv("car_malfunctions.csv")
df.info()
```

```python
xpoint=df["Age"]
y1point=df["malfunctions"]
y2point=df["DaysOff"]

plt.subplot(1,2,1)
plt.scatter(xpoint,y1point)
plt.xlabel("Age")
plt.ylabel("malfunctions")

plt.subplot(1,2,2)
plt.scatter(xpoint,y2point)

plt.xlabel("Age")
plt.ylabel("DaysOff")
plt.show()
```
