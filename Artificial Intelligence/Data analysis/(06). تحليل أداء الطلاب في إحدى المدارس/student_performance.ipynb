```python
df.rename(columns={
    'race/ethnicity':'race',
    'parental level of education':'parental_education',
     'test preparation course': 'preparation_course',
    'math score':'math_score',
     'reading score': 'reading_score',
     'writing score': 'writing_score'
}, inplace=True
)
```

```python
df.columns
```

```python
df.describe()
```

```python
reading_full=df[df["reading_score"]==100]['reading_score'].count()
writing_full=df[df["writing_score"]==100]['writing_score'].count()
math_full=df[df["math_score"]==100]['math_score'].count()
print(f'number of student with full mark in reading: {reading_full}')
print(f'number of student with full mark in writing: {writing_full}')
print(f'number of student with full mark in math: {math_full}')
```

```python
pass_grade = 50

df["pass_math"]= "yes"
filt= (df["math_score"]<pass_grade)
df.loc[filt,"pass_math"]="no"
df[["math_score","pass_math"]].head()

df["pass_reading"]= "yes"
filt= (df["reading_score"]<pass_grade)
df.loc[filt,"pass_reading"]="no"
df[["reading_score","pass_reading"]].head()

df["pass_writing"]= "yes"
filt= (df["writing_score"]<pass_grade)
df.loc[filt,"pass_writing"]="no"
df[["writing_score","pass_writing"]].head()
df
```

```python
pass_math_count= df.groupby("pass_math")["pass_math"].count()
pass_math_count
```

```python
pass_math_count= df.groupby("pass_reading")["pass_reading"].count()
pass_math_count
```

```python
pass_math_count= df.groupby("pass_writing")["pass_writing"].count()
pass_math_count
```

```python
fig , ax=plt.subplots(1,3, figsize=(16,4))
sns.histplot(x="math_score",data= df, ax=ax[0],color="red")
sns.histplot(x="reading_score",data= df, ax=ax[1],color="red")
sns.histplot(x="writing_score",data= df, ax=ax[2],color="red")

math_mean=df["math_score"].mean()
reading_score=df["reading_score"].mean()
writing_score=df["writing_score"].mean()
print(f' Average score in Math:{math_mean}\nAverage score in reading:{reading_score}\nAverage scoer in writing:{writing_score}\n\n')
plt.show()
```

```python
fig , ax=plt.subplots(1,3, figsize=(16,4))
sns.countplot(x="pass_math",data=df, ax=ax[0])
sns.countplot(x="pass_reading", data =df , ax= ax[1])
sns.countplot(x="pass_writing" , data =df ,ax= ax[2])

pass_math_count= df.groupby("pass_math")["pass_math"].count()
print(pass_math_count)

pass_reading_count= df.groupby("pass_reading")["pass_reading"].count()
print(pass_reading_count)

pass_writing_count= df.groupby("pass_writing")["pass_writing"].count()
print(pass_writing_count)
plt.show
```

```python
plt.rcParams['figure.figsize']=(40,20)
plt.rcParams['font.size']=23

plt.subplot(2,3,1)
size =df["gender"].value_counts()

labels= "Female","Male"
color =["red","green"]
plt.pie(size,colors=color,labels= labels,autopct="%.2f%%")
plt.title("Gender",fontweight='bold')

plt.subplot(2,3,2)
size= df["race"].value_counts()
labels="group C","group D","group B","group E","group A"
color =["red","green","blue","cyan","orange"]
plt.pie(size,colors=color,labels=labels,autopct="%.2f%%")
plt.title("Race",fontweight='bold')      

plt.subplot(2,3,3)
size = df["lunch"].value_counts()
labels="standard","free"
color =["orange","blue"]
plt.pie(size,colors=color,labels=labels,autopct="%.2f%%")
plt.title("Lunch",fontweight='bold')      

plt.subplot(2,3,4)
size = df["preparation_course"].value_counts()
labels = "none","completed"
color= ["blue","red"]
plt.pie(size,colors=color,labels=labels,autopct="%.2f%%")
plt.title("preparation",fontweight='bold')

plt.subplot(2,3,5)
size =df["parental_education"].value_counts()
labels = "bachelor degree","some college","master_degree","high school","associate","same high school"
color =["red","green","blue","cyan","orange","grey"]
plt.pie(size,colors=color,labels=labels,autopct="%.2f%%")
plt.title("parental_education",fontweight='bold')

plt.show()
```

```python
fig, (ax1,ax2,ax3)=plt.subplots(1,3)
fig.suptitle("Comparing Male & Femal scores for all three subjects:", fontsize=20)
sns.boxplot(x="gender",y="math_score",data=df,ax=ax1)
ax1.title.set_text("Math vs gender")

sns.boxplot(x="gender",y="reading_score",data=df,ax=ax2)
ax2.title.set_text("Reading vs gender")

sns.boxplot(x="gender",y="writing_score",data=df,ax=ax3)
ax3.title.set_text("Writing vs gender")
```
