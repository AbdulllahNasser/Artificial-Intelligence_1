```python
import pandas as pd
```

```python
df= pd.read_csv("Ecommerce.csv")
df.info()
```

```python
gc= df.groupby("Country")
gc.groups
               
```

```python
gc.get_group('KSA')
```

```python
gc= df.groupby(['Country','Gender'])
for name,gender in gc:
    print(name)
    print(gender)
```

```python
gc= df.groupby("Country")['Id'].count()
gc
```

```python
gc= df.groupby(['Country','Gender']).count()
gc
```

```python
gc= df.groupby("Country")['Amount'].mean()
gc
```
