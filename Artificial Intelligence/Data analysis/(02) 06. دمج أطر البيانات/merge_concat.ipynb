```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'column_a':[1,2,3,4],
    'column_b':['a','b','c','d'],
    'column_c':[True,True,False,False]
})
df1
```

```python
df2 = pd.DataFrame({
    'column_a':[4,7,3,4],
    'column_b':['U','G','E','d'],
    'column_c':[False,False,True,True]
})
df2
```

```python
df = pd.concat([df1,df2])
df
```

```python
df = pd.concat([df1,df2], keys=['df1','df2'])
df
```

```python
df = pd.concat([df1,df2], ignore_index=True)
df
```

```python
df = pd.concat([df1,df2], axis=1)
df
```
