```python
import pandas as pd 
df= pd.read_csv("Students_exams.csv")
df.info()
```

```python
df.columns
```

```python
df.rename(columns={
    'math score':'math_score',
     'reading score': 'reading_score',
    'writing score':'writing_score'
},inplace=True
         )
```

```python
df.columns
```

```python
statics_df= pd.DataFrame(columns=['statics','math','reading','writing'])
```

```python
x=df['math_score'].count()
y=df['reading_score'].count()
i=df['writing_score'].count()
row ={'statics':'count','math':x,'reading':y,'writing':i}
#statics_df=statics_df.append(row,ignoer_index=True)
 
statics_df = pd.concat(
    [statics_df, pd.DataFrame([row])],
    ignore_index=True)
```

```python
x=df['math_score'].min()
y=df['reading_score'].min()
i=df['writing_score'].min()
row ={'statics':'min','math':x,'reading':y,'writing':i}
#statics_df=statics_df.append(row,ignoer_index=True)
 
statics_df = pd.concat(
    [statics_df, pd.DataFrame([row])],
    ignore_index=True)
```
