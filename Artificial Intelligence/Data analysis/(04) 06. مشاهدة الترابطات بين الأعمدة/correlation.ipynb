```python
import pandas as pd
df= pd.read_csv("stdMarks.csv")
dfcorrel=df[["Math","Phy","Lang"]]
correlation=dfcorrel.corr(method='pearson')
correlation
```

```python
import matplotlib.pyplot as plt
```

```python
fig=  plt.figure()

ax= fig.add_subplot(1,1,1)

cax= ax.matshow(correlation,vmin=-1,vmax=1)

fig.colorbar(cax)

ticks= [0,1,2]

ax.set_xticks(ticks)
ax.set_yticks(ticks)

ax.set_xticklabels(["Math","Phy","Lange"])
ax.set_yticklabels(["Math","Phy","Lange"])

plt.show()
```
