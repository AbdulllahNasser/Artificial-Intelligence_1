```python
import matplotlib.pyplot as plt
```

```python
y=[36,25,24,12]
plt.pie(y)
plt.show()
```

```python
mylabels=['appels','banans','cherres','dated']
plt.pie(y, labels=mylabels)
plt.show()
```

```python
myexplode=[0.2,0,0,0]
plt.pie(y, labels=mylabels,explode=myexplode)
plt.show()
```

```python
plt.pie(y, labels=mylabels,explode=myexplode,shadow=True)
```

```python
plt.pie(y, labels=mylabels)
plt.legend()
```
