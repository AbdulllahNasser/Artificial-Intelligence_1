```python
import pandas as pd
```

```python
df1 =pd.DataFrame({'city':['A','B','C',],
                   'day1':[22,5,28],
                   'day2':[10,14,15],
                   'day3':[33,44,5],
                   'day4':[55,62,79],
                   'day5':[11,33,55]})
df1
```

```python
df1.melt(id_vars=['city'])
```

```python
df1.melt(id_vars=['city'], var_name='date',value_name='temprature')
```

```python
df1.melt(id_vars=['city'], var_name='date',value_name='temprature').sort_values(by='city')
```

```python
df1.melt(id_vars=['city'], var_name='date',value_name='temprature').sort_values(by='city').reset_index(drop=True)
```
